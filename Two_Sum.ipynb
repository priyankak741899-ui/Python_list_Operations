```python
numbers=[]
n=int(input("Enter the number of elements:"))
for i in range(n):
    num=int(input("Enter a number:"))
    numbers.append(num)
target=int(input("Enter the target:"))    
for j in range(len(numbers)):
    for k in range(j+1,len(numbers)):
        if(numbers[j]+numbers[k])==target:
            print([j,k])
            break
```
